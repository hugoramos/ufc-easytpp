```python
import numpy as np
import matplotlib.pyplot as plt

# Simple Poisson process simulation
np.random.seed(42)

lam = 1.2  # rate per unit time
T = 10.0   # observation window

# Simulate interarrival times ~ Exp(lam)
times = []
t = 0.0
while True:
    t += np.random.exponential(1/lam)
    if t > T:
        break
    times.append(t)

times = np.array(times)
n_events = len(times)

# Plot events on a time line
plt.figure(figsize=(10, 2.8))
plt.hlines(0, 0, T, linewidth=2)
plt.vlines(times, -0.3, 0.3, linewidth=2)  # event marks
plt.scatter(times, np.zeros_like(times), s=30)

plt.yticks([])
plt.xlim(0, T)
plt.ylim(-0.6, 0.6)
plt.xlabel("Time")
plt.title(f"Simple Poisson Process Realization (λ={lam}, T={T}, events={n_events})")

plt.show()
```
